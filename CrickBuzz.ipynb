```python
from selenium import webdriver
import time
driver=webdriver.Chrome()
url="https://www.cricbuzz.com/cricket-series/9241/indian-premier-league-2026/matches"
driver.get(url)
time.sleep(5) 
html=driver.page_source
with open("ipl.html", "w", encoding="utf-8") as f:
    f.write(html)
print("HTML File Saved")
driver.quit()
```

```python
from bs4 import BeautifulSoup
with open("ipl.html", "r", encoding="utf-8") as f:
    file = f.read()
soup = BeautifulSoup(file, "html.parser")
matches=soup.find_all("div",class_="mt-2")
team_1=[]
team_2=[]
m_number=[]
span=[]
div=[]
for i in matches:
    sp=i.find_all("span")
    span.append(sp)       
    
    if len(sp)==13:
        mn=sp[3]
        if mn:
            m_number.append(mn.text.split("•")[0].strip()[:-8])
        else:
            m_number.append("None")
        t1=sp[5]
        if t1:
            team_1.append(t1.text)
        else:
            team_1.append("None")
    
        t2=sp[8]
        if t2:
            team_2.append(t2.text)
        else:
            team_2.append("None")
            
    if len(sp)==24:
        mn1=sp[3]
        if mn:
            m_number.append(mn1.text.split("•")[0].strip()[:-8])
        else:
            m_number.append("None")
        t1=sp[5]
        if t1:
            team_1.append(t1.text)
        else:
            team_1.append("None")
        t11=sp[16]
        if t11:
            team_1.append(t11.text)
        else:
            team_1.append("None")

        mn2=sp[13]
        if mn:
            m_number.append(mn2.text.split("•")[0].strip()[:-8])
        else:
            m_number.append("None")
        t2=sp[8]
        if t2:
            team_2.append(t2.text)
        else:
            team_2.append("None")
        t22=sp[19]
        if t22:
            team_2.append(t22.text)
        else:
            team_2.append("None")
```

```python
import pandas as pd

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

data = {
    "Match No.":m_number,
    "Team 1": team_1,
    "Team 2": team_2,
    "Winner":winner,
    "Team 1 Run":team_1_run,
    "Team 2 Run":team_2_run,
    "Team 1 Wicket":team_1_wicket,
    "Team 2 Wicket":team_2_wicket,
    "Team 1 Played Ball":team_1_playedBall,
    "Team 2 Played Ball":team_2_playedBall
}
df = pd.DataFrame(data)
```

```python
teams=["CSK","RCB","MI","LSG","RR","DC","KKR","SRH","GT","PBKS"]
t_play=[]
t_win=[]
t_lose=[]
t_point=[]
t_cancleMatch=[]
t_runs_score=[]
t_over_face=[]
t_runs_conced=[]
t_over_throw=[]
net_run_rate=[]

int_colmn = ["Team 1 Run","Team 2 Run","Team 1 Wicket","Team 2 Wicket","Team 1 Played Ball","Team 2 Played Ball"]
for i in int_colmn:
    df[i] = df[i].astype(int)
    
for i in teams:
    t_play.append((df[df["Team 1"]==i]).shape[0]+(df[df["Team 2"]==i]).shape[0])
    t_win.append((df[df["Winner"]==i]).shape[0])
    t_lose.append((df[ ((df["Team 1"]==i) | (df["Team 2"]==i))& ((df["Winner"]!=i) & (df["Winner"]!="Cancle"))]).shape[0])
    t_cancleMatch.append(df[((df["Team 1"]==i) | (df["Team 2"]==i))&(df["Winner"]=="Cancle")].shape[0])
    run_score=(df[df["Team 1"]==i]["Team 1 Run"].sum()+df[df["Team 2"]==i]["Team 2 Run"].sum())
    t_runs_score.append(run_score)
    run_conced=(df[df["Team 1"]==i]["Team 2 Run"].sum()+df[df["Team 2"]==i]["Team 1 Run"].sum())
    t_runs_conced.append(run_conced)
    over_face=((df[(df["Team 1"]==i)&(df["Team 1 Wicket"]!=10)]["Team 1 Played Ball"].sum())+(df[(df["Team 2"]==i)&(df["Team 2 Wicket"]!=10)]["Team 2 Played Ball"].sum()))/6
    over_face+=((df[(df["Team 1"]==i)&(df["Team 1 Wicket"]==10)].shape[0])+(df[(df["Team 2"]==i)&(df["Team 2 Wicket"]==10)].shape[0]))*20
    t_over_face.append(round(over_face,2))
    over_throw=((df[(df["Team 1"]==i)&(df["Team 2 Wicket"]!=10)]["Team 2 Played Ball"].sum())+(df[(df["Team 2"]==i)&(df["Team 1 Wicket"]!=10)]["Team 1 Played Ball"].sum()))/6
    over_throw+=((df[(df["Team 1"]==i)&(df["Team 2 Wicket"]==10)].shape[0])+(df[(df["Team 2"]==i)&(df["Team 1 Wicket"]==10)].shape[0]))*20
    t_over_throw.append(round(over_throw,2))
    net_run_rate.append(round((run_score/over_face)-(run_conced/over_throw),3))
    
for i in range(len(t_win)):
    t_point.append((t_win[i]*2)+t_cancleMatch[i])
```

```python
point_table_dic={
    "Team":teams,
    "Played Match":t_play,
    "Win":t_win,
    "Lose":t_lose,
    "Cancled Match":t_cancleMatch,
    "Point":t_point,
    "Net Run Rate":net_run_rate
}
point_table_df=pd.DataFrame(point_table_dic)
point_table_df.sort_values(by=["Point","Net Run Rate"],ascending=[False,False],inplace=True)
point_table_df.set_index("Team", inplace=True)
print(point_table_df)
```
